> **Tip for a C# dev:** Python is dynamically typed, uses indentation instead of `{}`, and has no semicolons. Everything is an object, including functions and classes.

## 1. Variables and types
No declarations: a name is bound to an object, and the *object* carries the type. Type hints exist but are not enforced.

```python
age = 28                 # int   (C#: int)   - Python ints never overflow
salary = 1_250_000.50    # float (C#: double) - underscores are just for readability
name = "Shubham"         # str   (C#: string)
is_learning = True       # bool  (True/False capitalised)
nothing = None           # None  (C#: null)

for value in [age, salary, name, is_learning, nothing]:
    print(f"{value!r:>12} -> {type(value).__name__}")

print(2 ** 100)          # huge ints just work
print(7 / 2, 7 // 2, 7 % 2)   # true division, floor division, remainder
```

```python
# f-strings: Python's string interpolation (C#: $"...")
pi = 3.14159265
print(f"{name} is {age} years old; pi to 2 dp = {pi:.2f}; salary = {salary:,.0f}")

# Dynamic typing: the same name can point to a different type later
x = 10
x = "now I'm a string"
print(x)
```

## 2. Strings
Strings are **immutable** sequences. Slicing `s[start:stop:step]` works on every sequence type.

```python
s = "Machine Learning"
print(s[0], s[-1])        # first, last
print(s[0:7])             # 'Machine' (stop index is exclusive)
print(s[::-1])            # reversed
print(s.lower(), s.upper(), s.split(" "), "-".join(["a", "b", "c"]))
print("Learn" in s, s.replace("Machine", "Deep"), len(s))
```

## 3. Lists — ordered, mutable (C#: `List<T>`)
The workhorse. Can hold mixed types, but in ML you'll usually keep them homogeneous.

```python
marks = [78, 92, 65, 88]
marks.append(95)          # add to end
marks.insert(0, 70)       # insert at index
marks.remove(65)          # remove first matching value
last = marks.pop()        # remove and return last
print(marks, "popped:", last)
print(sorted(marks), sorted(marks, reverse=True), max(marks), sum(marks) / len(marks))
```

```python
# PITFALL: assignment does NOT copy. Both names point to the same list object.
a = [1, 2, 3]
b = a
b.append(99)
print("a after changing b:", a)       # a changed too!

c = a.copy()                          # or list(a) or a[:]
c.append(100)
print("a after changing a copy:", a)  # unchanged
```

## 4. Tuples — ordered, immutable
Use for fixed records and for returning multiple values. *Unpacking* is everywhere in Python.

```python
point = (3, 4)
x, y = point                       # unpacking
print(x, y)

a, b = 1, 2
a, b = b, a                        # swap without a temp variable
print(a, b)

first, *rest = [10, 20, 30, 40]    # star-unpacking
print(first, rest)
```

## 5. Dictionaries — key → value (C#: `Dictionary<K,V>`)
Keys must be immutable (str, int, tuple). Insertion order is preserved.

```python
student = {"name": "Asha", "roll": 101, "marks": {"math": 91, "physics": 84}}
print(student["name"])
print(student.get("email", "no email"))   # safe lookup with default (no KeyError)
student["email"] = "asha@example.com"     # add or update
for key, value in student.items():
    print(f"{key:>6}: {value}")
print(student["marks"]["math"])           # nested access
```

## 6. Sets — unique, unordered (C#: `HashSet<T>`)
Great for de-duplication and fast membership tests.

```python
ml_students = {"Asha", "Ravi", "Meena", "John"}
dl_students = {"Ravi", "John", "Priya"}
print(ml_students & dl_students)    # intersection: in both
print(ml_students | dl_students)    # union
print(ml_students - dl_students)    # only in ML
print(set([1, 1, 2, 2, 3]))         # de-duplicate
```

## 7. Control flow
Indentation (4 spaces) defines blocks. `enumerate` and `zip` replace most index-based loops.

```python
score = 76
if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
else:
    grade = "C"
print("grade:", grade)

names = ["Asha", "Ravi", "Meena"]
scores = [91, 67, 84]
for i, (n, sc) in enumerate(zip(names, scores), start=1):
    print(i, n, sc, "PASS" if sc >= 70 else "FAIL")   # inline if = C# ternary

n = 0
while n < 3:
    n += 1          # no n++ in Python
print("loop ended at", n)
```

## 8. Comprehensions — Python's LINQ
| C# LINQ | Python |
|---|---|
| `nums.Select(x => x*x)` | `[x*x for x in nums]` |
| `nums.Where(x => x>2)` | `[x for x in nums if x > 2]` |
| `ToDictionary(...)` | `{k: v for k, v in pairs}` |

```python
nums = [1, 2, 3, 4, 5, 6]
squares = [x * x for x in nums]
evens = [x for x in nums if x % 2 == 0]
labels = ["even" if x % 2 == 0 else "odd" for x in nums]
square_map = {x: x * x for x in nums}
unique_lengths = {len(w) for w in ["ml", "ai", "data", "model"]}
print(squares, evens, labels, square_map, unique_lengths, sep="\n")

# Nested comprehension: flatten a 2-D list (a matrix)
matrix = [[1, 2, 3], [4, 5, 6]]
print([v for row in matrix for v in row])
```

## 9. Functions
Default values, keyword arguments, multiple return values, docstrings and type hints.

```python
def describe(values: list[float], round_to: int = 2) -> tuple[float, float, float]:
    """Return (min, max, mean) of a list of numbers."""
    mean = sum(values) / len(values)
    return min(values), max(values), round(mean, round_to)

lo, hi, avg = describe([72, 88, 95, 61])
print(lo, hi, avg)
print(describe([1, 2, 2], round_to=3))   # keyword argument
help(describe)
```

```python
# *args collects extra positional args into a tuple; **kwargs collects keyword args into a dict
def train_model(model_name, *datasets, **hyperparams):
    print("model:", model_name)
    print("datasets:", datasets)
    print("hyperparams:", hyperparams)

train_model("RandomForest", "train.csv", "extra.csv", n_estimators=200, max_depth=8)

# Unpacking into a call: very common when passing config dicts to ML libraries
config = {"n_estimators": 100, "max_depth": 5}
train_model("XGBoost", **config)
```

```python
# PITFALL: never use a mutable default argument. It is created ONCE and shared across calls.
def add_bad(item, bucket=[]):
    bucket.append(item)
    return bucket

def add_good(item, bucket=None):
    if bucket is None:
        bucket = []
    bucket.append(item)
    return bucket

print(add_bad(1), add_bad(2))     # [1, 2] [1, 2]  <- surprise!
print(add_good(1), add_good(2))   # [1] [2]
```

## 10. Lambdas, `sorted`, `map`, `filter`
A lambda is an anonymous one-expression function (C#: `x => x * 2`). Its most common use is as a `key=` for sorting.

```python
students = [("Asha", 91), ("Ravi", 67), ("Meena", 84)]
print(sorted(students, key=lambda s: s[1], reverse=True))   # sort by marks, descending
print(list(map(lambda s: s[0].upper(), students)))
print(list(filter(lambda s: s[1] >= 80, students)))
# Prefer comprehensions for map/filter. They are more readable:
print([name for name, m in students if m >= 80])
```

## 11. Errors (C#: try/catch/finally)

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("cannot divide by zero")
        return None
    finally:
        print("finally always runs")

print(safe_divide(10, 2))
print(safe_divide(1, 0))

def set_learning_rate(lr):
    if not 0 < lr < 1:
        raise ValueError(f"learning rate must be in (0, 1), got {lr}")
    return lr

try:
    set_learning_rate(5)
except ValueError as e:
    print("caught:", e)
```

## 12. Classes and dataclasses
`self` is C#'s `this`, made explicit. `__init__` is the constructor. `@dataclass` generates `__init__` and `__repr__` for you, like a C# `record`.

```python
class LinearModel:
    """A tiny y = w*x + b model, the same shape as every sklearn estimator."""
    def __init__(self, w=0.0, b=0.0):
        self.w = w
        self.b = b

    def predict(self, x):
        return self.w * x + self.b

    def __repr__(self):
        return f"LinearModel(w={self.w}, b={self.b})"

m = LinearModel(w=2, b=1)
print(m, [m.predict(x) for x in range(4)])
```

```python
from dataclasses import dataclass, field

@dataclass
class Experiment:
    name: str
    accuracy: float
    tags: list[str] = field(default_factory=list)   # safe mutable default

    @property
    def is_good(self) -> bool:
        return self.accuracy >= 0.85

exps = [Experiment("baseline", 0.78), Experiment("tuned-rf", 0.88, ["rf"])]
print(exps)
print([e.name for e in exps if e.is_good])
```

# 14. LIVE DEMO — Student Marks Analyzer (core Python only)
Real ML projects start exactly like this: **load raw data → clean it → model the records → compute statistics → report**.
Tomorrow we'll redo this in 5 lines of Pandas, and you'll appreciate what Pandas does for you.

**Step 1 — Raw data.** It is deliberately messy: a missing value, a duplicate row, inconsistent name casing and an out-of-range mark.

```python
import csv, io

RAW_CSV = """name,roll,math,physics,chemistry,english
Asha,101,91,84,88,79
ravi,102,67,72,,81
Meena,103,84,90,93,88
John,104,45,58,52,70
Priya,105,99,95,97,92
Ravi,102,67,72,,81
Arjun,106,73,64,70,105
Fatima,107,88,79,85,90
"""

rows = list(csv.DictReader(io.StringIO(RAW_CSV)))
print(f"{len(rows)} raw rows")
print(rows[1])
```

**Step 2 — Model each record** as a dataclass, with derived properties.

```python
from dataclasses import dataclass

SUBJECTS = ["math", "physics", "chemistry", "english"]

@dataclass
class Student:
    name: str
    roll: int
    marks: dict[str, float]

    @property
    def average(self) -> float:
        return round(sum(self.marks.values()) / len(self.marks), 2)

    @property
    def grade(self) -> str:
        return grade_for(self.average)

def grade_for(avg: float) -> str:
    bands = [(90, "A+"), (80, "A"), (70, "B"), (60, "C"), (50, "D")]
    for cutoff, letter in bands:
        if avg >= cutoff:
            return letter
    return "F"
```

**Step 3 — Clean the data**: drop duplicates, normalise names, cap invalid marks and impute missing marks with the subject mean. These are the same decisions you'll make on every real dataset.

```python
import statistics as st

def clean(rows: list[dict]) -> list[Student]:
    seen_rolls, report = set(), []
    unique = []
    for r in rows:                                  # 1) de-duplicate by roll number
        if r["roll"] in seen_rolls:
            report.append(f"dropped duplicate roll {r['roll']}")
            continue
        seen_rolls.add(r["roll"])
        unique.append(r)

    subject_means = {                               # 2) means from valid values only
        s: st.mean(float(r[s]) for r in unique if r[s] and 0 <= float(r[s]) <= 100)
        for s in SUBJECTS
    }

    students = []
    for r in unique:
        marks = {}
        for s in SUBJECTS:
            if r[s] == "":                          # 3) impute missing
                marks[s] = round(subject_means[s], 1)
                report.append(f"{r['name']}: imputed {s} = {marks[s]}")
            elif float(r[s]) > 100:                 # 4) cap out-of-range
                marks[s] = 100.0
                report.append(f"{r['name']}: capped {s} {r[s]} -> 100")
            else:
                marks[s] = float(r[s])
        students.append(Student(r["name"].strip().title(), int(r["roll"]), marks))
    return students, report

students, cleaning_log = clean(rows)
print(*cleaning_log, sep="\n")
print(f"\n{len(students)} clean students")
```

**Step 4 — Analyse**: rank, subject statistics, toppers and grade distribution, all with comprehensions, `sorted`, `max` and `Counter`.

```python
from collections import Counter

ranked = sorted(students, key=lambda s: s.average, reverse=True)

print(f"{'Rank':<5}{'Name':<8}{'Roll':<6}{'Avg':>6}  Grade")
for i, s in enumerate(ranked, 1):
    print(f"{i:<5}{s.name:<8}{s.roll:<6}{s.average:>6}  {s.grade}")

print("\nSubject statistics")
for subj in SUBJECTS:
    vals = [s.marks[subj] for s in students]
    topper = max(students, key=lambda s: s.marks[subj])
    print(f"  {subj:<10} mean={st.mean(vals):6.2f}  std={st.stdev(vals):5.2f}  "
          f"topper={topper.name} ({topper.marks[subj]:.0f})")

print("\nGrade distribution")
dist = Counter(s.grade for s in students)
for g in ["A+", "A", "B", "C", "D", "F"]:
    print(f"  {g:<3}{'█' * dist.get(g, 0)} {dist.get(g, 0)}")
```

**Step 5 — Flag at-risk students** with a simple *rule-based classifier*. On Day 9 you'll replace hand-written rules like this with a learned logistic-regression model. That swap is the whole idea of machine learning.

```python
def at_risk(s: Student, threshold: float = 60) -> bool:
    weak_subjects = [subj for subj, m in s.marks.items() if m < threshold]
    return s.average < 65 or len(weak_subjects) >= 2

flagged = [(s.name, s.average) for s in students if at_risk(s)]
print("At-risk students:", flagged)
```
